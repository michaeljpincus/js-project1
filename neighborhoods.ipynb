```python
# Adding ISO2 codes so I can have flags in the javascript map

import pycountry

overrides = {
    "Dominican Republic": "DO",
    "Korea, South": "KR",
    "South Korea": "KR",
    "Trinidad & Tobago": "TT",
    "Trinidad and Tobago": "TT",
    "Cabo Verde": "CV",
    "Ivory Coast": "CI",
    "Cote d'Ivoire": "CI",
    "Bosnia & Herzegovina": "BA",
    "St. Vincent and the Grenadines": "VC",
    "Vietnam": "VN",
    "Russia": "RU",
    "Laos": "LA",
    "Taiwan": "TW",
    "Hong Kong": "HK",
    "Macedonia": "MK",
    "Venezuela": "VE",
    "Bolivia": "BO",
    "Iran": "IR",
    "Syria": "SY",
    "Moldova": "MD",
    "Czech Republic": "CZ",
    "Congo (Kinshasa)": "CD",
    "Congo (Brazzaville)": "CG",
    "Democratic Republic of Congo": "CD",
    "Burma": "MM",
    "Myanmar": "MM",
    "Korea, South": "KR",
    "South Korea": "KR",
    "Korea": "KR"
}

def name_to_iso2(name):
    if pd.isna(name) or name.strip() == "":
        return None
    if name in overrides:
        return overrides[name]
    try:
        return pycountry.countries.search_fuzzy(name)[0].alpha_2
    except LookupError:
        return None

for i in range(1,6):
    final_joined[f'Top {i} ISO2'] = final_joined[f'Top {i} Country'].apply(name_to_iso2)
```
